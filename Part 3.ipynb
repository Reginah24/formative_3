# Bayesian Probability Analysis of IMDb Movie Reviews

## Introduction

This project applies Bayesian Probability to IMDb movie reviews to estimate the probability that a review is positive based on selected keywords.

The dataset contains 50,000 movie reviews labelled as either positive or negative.

The analysis uses Prior, Likelihood, Marginal and Posterior probabilities to understand the relationship between selected keywords and movie-review sentiment.

## 1. Loading the IMDb Dataset

## 2. Selecting Keywords

We selected four positive keywords and four negative keywords to analyse their relationship with movie-review sentiment.

```python
positive_keywords = [
    "amazing",
    "excellent",
    "wonderful",
    "enjoyable"
]

negative_keywords = [
    "terrible",
    "boring",
    "awful",
    "disappointing"
]

print("Positive keywords:", positive_keywords)
print("Negative keywords:", negative_keywords)
```

## 3. Counting Keyword Occurrences

We count how many positive and negative reviews contain each selected keyword. This helps us understand how frequently each word appears in both sentiment categories.

```python
# Combine all selected keywords
all_keywords = positive_keywords + negative_keywords

# Count reviews containing each keyword
for word in all_keywords:
    positive_count = df[
        (df['sentiment'] == 'positive') &
        (df['review'].str.lower().str.contains(word, regex=False))
    ].shape[0]

    negative_count = df[
        (df['sentiment'] == 'negative') &
        (df['review'].str.lower().str.contains(word, regex=False))
    ].shape[0]

    print(f"\nKeyword: {word}")
    print("Positive reviews:", positive_count)
    print("Negative reviews:", negative_count)
```

## 4. Calculating Prior Probabilities

Prior probability represents the probability of selecting a positive or negative review before considering any keywords.

We calculate it by dividing the number of reviews in each category by the total number of reviews.

```python
total_reviews = len(df)

positive_total = len(df[df['sentiment'] == 'positive'])
negative_total = len(df[df['sentiment'] == 'negative'])

prior_positive = positive_total / total_reviews
prior_negative = negative_total / total_reviews

print("Total reviews:", total_reviews)
print("Positive reviews:", positive_total)
print("Negative reviews:", negative_total)

print("Prior Positive:", prior_positive)
print("Prior Negative:", prior_negative)
```

## 5. Calculating Likelihood, Marginal and Posterior Probabilities

- **Likelihood:** The probability of finding a keyword in a positive review.
- **Marginal:** The overall probability of finding that keyword in any review.
- **Posterior:** The probability that a review is positive given that the selected keyword appears.

We use Bayes' Theorem to calculate the posterior probability:

P(Positive | Keyword) = [P(Keyword | Positive) × P(Positive)] / P(Keyword)

```python
results = []

for word in all_keywords:
    positive_count = df[
        (df['sentiment'] == 'positive') &
        (df['review'].str.lower().str.contains(word, regex=False))
    ].shape[0]

    negative_count = df[
        (df['sentiment'] == 'negative') &
        (df['review'].str.lower().str.contains(word, regex=False))
    ].shape[0]

    likelihood = positive_count / positive_total

    marginal = (positive_count + negative_count) / total_reviews

    posterior = (likelihood * prior_positive) / marginal

    results.append({
        'Keyword': word,
        'Likelihood': likelihood,
        'Marginal': marginal,
        'Posterior': posterior
    })

probability_table = pd.DataFrame(results)

print(probability_table.round(4))
```

## 6. Results and Interpretation

The table below presents the calculated Likelihood, Marginal and Posterior probabilities for all eight selected keywords.

A higher Posterior probability indicates a stronger association with positive reviews, while a lower probability indicates a stronger association with negative reviews.

```python
for word in all_keywords:
    row = probability_table[probability_table['Keyword'] == word].iloc[0]

    print(f"{word}:")
    print("Posterior using Bayes:", round(row['Posterior'], 4))
```
